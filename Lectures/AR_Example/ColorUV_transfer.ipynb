Demo on using UV mapping to transfer image/texture onto 3D objects

Textured 3D meshes have:
    vertices- the XYZ point of each node
    faces/tris-  the connections of the index for the 3 (or 4) corners of connected vertices
    UV map- the mapping of each node into an texture image
    texture image

The texture for a 3D object is done by
1) Find the 2D image points for the three points corresponding to a face
        if XYZ is in normalized space, then this becomes
            i_normalized = X_normalized*image_width
            j_normalized = Y_normalized*image_height
2) In the texture image, the texture to display on the mesh in this face is
    defined by the three UV corrdinates.  Extract this triangular ROI image
3) Warp the triangular ROI image from the texture image to the (skewed) triangular 
    ROI defined by 2D projection of the nodes to image space


```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import open3d as o3d
import os
import math
```

```python
# Define the paths to your mesh and texture files
mesh_path = os.path.relpath("Pumpkin.obj")  
texture_path = os.path.relpath("Pumpkin.JPG") 

# 1. Load the 3D mesh
mesh = o3d.io.read_triangle_mesh(mesh_path)

pumpkin_img=cv2.cvtColor(cv2.imread(texture_path),cv2.COLOR_BGR2RGB)
```

```python
# Grab the vertices and faces info from the mesh
verts=np.asarray(mesh.vertices)
tris =np.asarray(mesh.triangles)
uvs=np.asarray(mesh.triangle_uvs)

# mean subtract and normalize the positions

 # make sure the object fits inside our 1x1x1 box
verts[:,0]=verts[:,0]-np.mean(verts,axis=0)[0]
verts[:,1]=verts[:,1]-np.mean(verts,axis=0)[1]
verts[:,2]=verts[:,2]-np.mean(verts,axis=0)[2]

verts=verts/np.max(np.abs(verts))/4
verts[:,0]=verts[:,0] # move to the center of the box
verts[:,1]=verts[:,1] 
verts[:,2]=verts[:,2] 
```

```python
theta = np.array([90,0,0]).astype(np.float32)*math.pi/180.0

RotMtxX = np.array([[1,0,0],
                    [0,math.cos(theta[0]),-math.sin(theta[0])],
                    [0,math.sin(theta[0]),math.cos(theta[0])]]) 
RotMtxY = np.array([[math.cos(theta[1]),0,math.sin(theta[1])],
                    [0,1,0],
                    [-math.sin(theta[1]),0,math.cos(theta[1])]]) 
RotMtxZ = np.array([[math.cos(theta[0]),-math.sin(theta[0]),0],
                    [math.sin(theta[0]),math.cos(theta[0]),0],
                    [0,0,1]]) 
R=RotMtxX@RotMtxY@RotMtxZ
```

```python
h=1080; w=1080
img = 255*np.ones((h,w,3)).astype(np.uint8)

verts_rot=verts@R

verts2D = np.zeros((verts_rot.shape[0],2)).astype(verts_rot.dtype)
verts2D[:,0]=verts_rot[:,0]*w+w/2
verts2D[:,1]=verts_rot[:,2]*h+h/2
```

```python

thickness=1
color=[0,255,0]
size=img.shape

idx=np.argsort(np.mean(verts_rot[tris,2],axis=1))

for idx,tri in enumerate(tris[idx,:]):
    pts=verts2D[tri,:]
    pts2=uvs[[idx*3,idx*3+1,idx*3+2],:]
    pts2[0,:]=pts2[0,:]*w
    pts2[1,:]=pts2[1,:]*h

    warpMat = cv2.getAffineTransform(np.float32(pts),np.float32(pts2))
  
    pts3=np.int64(pts)
    cv2.line(img,pts3[0,:], pts3[1,:],color,thickness=thickness)
    cv2.line(img,pts3[1,:], pts3[2,:],color,thickness=thickness)
    cv2.line(img,pts3[2,:], pts3[0,:],color,thickness=thickness)

plt.imshow(img)
```

```python
# Apply affine transform calculated using srcTri and dstTri to src and
# output an image of size.
def applyAffineTransform(src, srcTri, dstTri, size):

  # Given a pair of triangles, find the affine transform.
  warpMat = cv2.getAffineTransform(np.float32(srcTri), np.float32(dstTri))

  # Apply the Affine Transform just found to the src image
  dst = cv2.warpAffine(src, warpMat, (size[0], size[1]), None,
             flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_REFLECT_101)

  return dst

# Warps and alpha blends triangular regions from img1 and img2 to img
def warpTriangle(img1, img2, t1, t2):

  r1 = cv2.boundingRect(np.float32([t1]))
  r2 = cv2.boundingRect(np.float32([t2]))

  # Offset points by left top corner of the respective rectangles
  t1Rect = []
  t2Rect = []
  t2RectInt = []

  for i in range(0, 3):
    t1Rect.append(((t1[i][0] - r1[0]), (t1[i][1] - r1[1])))
    t2Rect.append(((t2[i][0] - r2[0]), (t2[i][1] - r2[1])))
    t2RectInt.append(((t2[i][0] - r2[0]), (t2[i][1] - r2[1])))

  # Get mask by filling triangle
  mask = np.zeros((r2[3], r2[2], 3), dtype=np.float32)
  mask=cv2.fillConvexPoly(mask, np.int32(t2RectInt), (1.0, 1.0, 1.0), 16, 0)

  h,w,_=mask.shape
  img2Rect = applyAffineTransform(img1, t1,t2,(w,h))

  img2Rect = img2Rect * mask

  # Copy triangular region of the rectangular patch to the output image
  img2[r2[1]:r2[1]+r2[3], r2[0]:r2[0]+r2[2]] = img2[r2[1]:r2[1]+r2[3], r2[0]:r2[0]+r2[2]] * ((1.0, 1.0, 1.0) - mask)
  img2[r2[1]:r2[1]+r2[3], r2[0]:r2[0]+r2[2]] = img2[r2[1]:r2[1]+r2[3], r2[0]:r2[0]+r2[2]] + img2Rect

  return
```

```python
idx=np.argsort(np.mean(verts_rot[tris,2],axis=1))

img = 255*np.ones((h,w,3)).astype(np.uint8)

hp,wp,_=pumpkin_img.shape

for idx,tri in enumerate(tris[idx,:]):
    pts=verts2D[tri,:]
    pts2=uvs[[idx,idx+1*tris.shape[0],idx+2*tris.shape[0]],:]
    pts2[:,0]=pts2[:,0]*hp
    pts2[:,1]=pts2[:,1]*wp
    
    warpTriangle(pumpkin_img, img, pts2, pts)
```
